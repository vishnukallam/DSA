# Sorting Algorithms

#### Bubble Sort

```python
''' 
The bubble sort algorithm is used to sort the elements in such a way that the elements are swapped by checking one by one
'''
# Algorithm:-
# step-1: Compare the adjacent (side by side) elements in the given array. 
#         If element-1 is greater than element-2 then swap those two.
# step-2: If the elements is not greater than the next elemnent then move the pointer.
# step-3: Repeate the process until the array is sorted.
# The complexity of the bubble sort is O(n**2)

arr = list(map(int, input("enter array: ").split()))
for i in range(0, len(arr)-1):
    for j in range(0, len(arr)-1-i):
        if arr[j]>arr[j+1]:
            print(arr)
            arr[j],arr[j+1] = arr[j+1],arr[j]
print(arr)
```

#### Insertion Sort

```python
'''
Insertion Sort is a sorting technique that is used to sort the elements using 2 sublist (which are sorted and unsorted).
The loop will start from the second element in the array. 
Store the second element in the Temp variable and compare the temp with first element of the array.
'''

arr = list(map(int, input().split()))
for i in range(1, len(arr)):
    temp =arr[i]
    j = i-1
    while j >= 0 and arr[j] > temp:
        arr[j+1] = arr[j]
        j -= 1
    arr[j+1] = temp
print(arr)
```

#### Selection Sort

```python
'''
Selection Sort is as simple as Insertion Sort. Same as insertion sort, it has 2 sublists (which are sorted and unsorted).
We have to find the minimum element in the given array and swap it with that minimum element.
By keeping the ele-1, which is in the 0th position, and comparing with other elements, and swapping them.
For selection sort, the number of iterations/passes is n-1.
This sorting technique required 2 loops 1 is for passes, and the other 1 is for swapping.
'''
arr = list(map(int, input().split()))
for i in range(0, len(arr)-1):
    mini = i
    for j in range(i+1, len(arr)):
        if arr[j] < arr[mini]:
            mini = j
    if mini != 0:
        print(f"If2 \n{arr}")
        arr[i], arr[mini] = arr[mini], arr[i]
print(arr)
```
